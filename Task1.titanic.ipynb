## We will Start with import the Pandas and numpy library to connect the dataset to the Machine algorithm Algorithm

```python
import pandas as pd
import numpy as np

#Load the CSV
df = pd.read_csv("tested.csv")

#overview
df.info()
df.head()
```

#### We can see there are many unnecessary things like:- Name,Ticket, Cabin ,PassengerId.... So we will drop it 

```python
df.drop(['PassengerId', 'Name', 'Ticket', 'Cabin'], axis=1, inplace=True)
```

```python
df
```

### we Have some missing data to handel 

```python
#Fill missing Age with median
df['Age'].fillna(df['Age'].median(), inplace=True)
```

### Fill missing data in Fare and Embarked

```python
df['Fare'].fillna(df['Fare'].median(), inplace=True)

## We have Embarked in a categorical column (like:-'C','Q','S') So we can't use 'median' and 'mean' so we use '.mode()' to returns the most frequent value in a column.
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
```

```python
df
```

#### We will use Label encoding for 'Sex' but 'Embarked' has more than two categories ('S','C','Q'),then use One-Hot Encoding to Convert it . 

```python
## using Sklearn for Label Encoding 
from sklearn.preprocessing import LabelEncoder
label_encoder_gender = LabelEncoder()
df['Sex'] = label_encoder_gender.fit_transform(df['Sex'])
df
```

```python
## using Sklearn for Ohe hot Encoding 
from sklearn.preprocessing import OneHotEncoder
onehot_Embarked = OneHotEncoder()
OH_Embarked = onehot_Embarked.fit_transform(df[['Embarked']]).toarray()
OH_Embarked
```

```python
Embarked_df = pd.DataFrame(OH_Embarked,columns=onehot_Embarked.get_feature_names_out(['Embarked']))
Embarked_df
```

### Combine one hot encoder columns with the original data

```python
df=pd.concat([df.drop('Embarked',axis=1),Embarked_df],axis=1)
df.head()
```

## Now we will Split Dependent and Independent variable 

```python
X = df.drop('Survived',axis=1)
y = df['Survived']
```

```python
X
```

```python
y
```

```python
df.shape
```

## Spliting The Data into Test set and Training Set

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25, random_state=42)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

## Featuring Scaling 

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
X_test
```

## Using LogisticRegression , RandomForestClassifier , DecisionTreeClassifier


```python
df
```

## LoogisticRegression

```python
from sklearn.linear_model import LogisticRegression
LR = LogisticRegression(random_state = 0)
LR.fit(X_train, y_train)
```

#### Predicting the test set 


```python
y_pred = LR.predict(X_test)
y_test_np = y_test.values
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test_np.reshape(len(y_test),1)),1))
```

### Accuracy and confusion matrix of LogisticRegression

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
RF = RandomForestClassifier(n_estimators = 10, criterion = 'entropy', random_state = 0)
RF.fit(X_train, y_train)
```

### Accuracy and confusion matrix of RandomForestClassifier

```python
from sklearn.metrics import confusion_matrix, accuracy_score
y_pred = RF.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```

## DecisionTreeClassifier

```python
from sklearn.tree import DecisionTreeClassifier
DT = DecisionTreeClassifier(criterion = 'entropy', random_state = 0)
DT.fit(X_train, y_train)
```

### Accuracy and confusion matrix of DecisionTreeClassifier

```python
from sklearn.metrics import confusion_matrix, accuracy_score
y_pred = DT.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```
